###Why to use matrix multiplication inbuild instead of doing it by hand:

```python
%%time
big_tensor= torch.rand(10000,10000)
```

```python
%%time
big_tensor = torch.rand(10000, 10000,
                                         dtype=torch.float32, # or torch.float16, torch.long, etc.
                                         device='cuda',        # 'cpu' or 'cuda' if a GPU is available
                                         requires_grad=False) # True if you want to compute gradients for this tensor
```

```python
#Matrix multiplication by hand:
%%time
value=0
for i in range(len(big_tensor)):
  value+= big_tensor[i]* big_tensor[i]
value
```

```python
#Matrix multiplication using pytorch inbuild function:
%%time
torch.matmul(big_tensor, big_tensor)
```

##Shape Errors

```python
TENSORA= torch.tensor([[1,2],
                       [2,3],
                       [3,4]])
TENSORB= torch.tensor([[5,6],
                       [6,7],
                       [7,8]])

torch.mm(TENSORA, TENSORB) #.mm means matmul (alias for matmul)
```

##Transpose
####To fix the upper error, due to the shape of the matrix.

```python
torch.mm(TENSORA, TENSORB.T) #TENSORB.T --> here .T means transposing the TENSORB
```

##Reshaping, stacking, squeezing and unsqueezing tensors:
* Reshaping- Reshapes an input tensor to a defined shape so that we can perform our desired mathematically operations on it.
* View- Return a view of an input tensor of certain shape but keep the same memory as the original tensor.
* Stacking- Combine multiple tensors on top of each other.
* Squeeze- removes all '1' dimensions from a tensor
* Unsqeeze- add a '1' dimension to a target tensor
* Permute- Return a view of the input with dimensions permuted (swapped) in a certain way.

```python
#Create a tensor:
x= torch.arange(1.,10.)
y= torch.arange(1., 10.)
x, x.shape, y, y.shape
```

```python
#Add an extra dimension:
x_reshaped= x.reshape(1,9) #here shape of x is 9 so values in our bracket could be anything until there multiplication =9, here 1*9=9, we can also write (3,3), since 3*3=9

#same multiplication condition also goes with y
y_transpose_reshaped= y.reshape(9,1)
x_reshaped, x_reshaped.shape, y_transpose_reshaped, y_transpose_reshaped.shape
```

```python
x_reshaped==y_transpose_reshaped
```

```python
x_reshaped==y_transpose_reshaped.T
```

```python
#Stack Tensors on top of each other:
x_stacked= torch.stack([x,x,x,x], dim=0)
x_stacked
```

```python
# Torch.squeeze:
x_reshaped
```

```python
x_reshaped.shape
```

```python
x_squeezed=x_reshaped.squeeze()
x_squeezed
```

```python
x_reshaped.squeeze().shape
```

```python
#Torch.unsqueeze:
x_unsqueezed= x_squeezed.unsqueeze(dim=1)
x_unsqueezed
```

```python
#Torch.permute- rearranges the dimension of the target tensor in a specified order:
x_original= torch.rand(size=(224,224,3))
x_original
```

```python
#permuting the original tensor x_original:
x_permuted= x_original.permute(2,0,1)
print(f"Original shape: {x_original.shape}")
print(f"Permuted shape: {x_permuted.shape}")
```
