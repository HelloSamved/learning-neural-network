###Random tensors

##Random Tensors
####Why Random Tensors?
####Random tensors are important because they way neural network learn is that they start with tensors full numbers and then adjust those random numbers to better represent the data.

###The flow looks like:
####Start with random numbers in tensor --> look at data --> update random numbers --> look at data --> (loop)

```python
#Create a random tensor of size:
random_tensor1= torch.rand(1,10,10)
random_tensor1
```

```python
random_tensor1.shape
```

```python
random_tensor2= torch.rand(4,10,10)
random_tensor2
```

```python
random_tensor2.shape, random_tensor2.ndim
```

```python
#Random tensor with similar shape to an image:
random_image_size_tensor= torch.rand(size=(3,224,224))
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

##Creating a range of tensors and tensors like

```python
#Use torch.arange()
one_to_ten=torch.arange(start=1, end= 100, step=5)
one_to_ten
```

```python
one_to_ten.shape, one_to_ten.ndim
```

```python
#Creating tensors like:
ten_zeros= torch.zeros_like(input=one_to_ten)
ten_zeros
```

##Tensor Data Types

```python
#Float 32 Type:
float_32_tensor= torch.tensor([1.0,2.0,3.0],
                               dtype=None, #this defines the data type of the tensor, here none will set it dtype as float32 by default
                               device= "cuda", #
                               requires_grad=False)
float_32_tensor, float_32_tensor.dtype
```

```python
float_16_tensor= float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
#Just Experimenting
x= float_32_tensor* float_16_tensor
x
```
