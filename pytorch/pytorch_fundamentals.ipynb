###Random tensors

##Random Tensors
####Why Random Tensors?
####Random tensors are important because they way neural network learn is that they start with tensors full numbers and then adjust those random numbers to better represent the data.

###The flow looks like:
####Start with random numbers in tensor --> look at data --> update random numbers --> look at data --> (loop)

```python
#Create a random tensor of size:
random_tensor1= torch.rand(1,10,10)
random_tensor1
```

```python
random_tensor1.shape
```

```python
random_tensor2= torch.rand(4,10,10)
random_tensor2
```

```python
random_tensor2.shape, random_tensor2.ndim
```

```python
#Random tensor with similar shape to an image:
random_image_size_tensor= torch.rand(size=(3,224,224))
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

###Matrix Multiplication:
#####Two main ways for this multiplication
1.   Element-wise multiplication
2.   Matrix multiplication



```python
#Element-wise multiplication:
print(TENSOR1, "*", TENSOR1)
print(f"Equals: {TENSOR1 * TENSOR1}")
```

###Why to use matrix multiplication inbuild instead of doing it by hand:

```python
%%time
big_tensor= torch.rand(10000,10000)
```

```python
%%time
big_tensor = torch.rand(10000, 10000,
                                         dtype=torch.float32, # or torch.float16, torch.long, etc.
                                         device='cuda',        # 'cpu' or 'cuda' if a GPU is available
                                         requires_grad=False) # True if you want to compute gradients for this tensor
```

```python
#Matrix multiplication by hand:
%%time
value=0
for i in range(len(big_tensor)):
  value+= big_tensor[i]* big_tensor[i]
value
```

```python
#Matrix multiplication using pytorch inbuild function:
%%time
torch.matmul(big_tensor, big_tensor)
```
